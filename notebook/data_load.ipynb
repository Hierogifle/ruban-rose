```python
import os
import shutil
import kagglehub

dossier_final = "../data/breast_histopathology" 

if not os.path.exists(dossier_final):
    print("📥 Téléchargement du dataset Breast Histopathology Images...")
    os.environ["KAGGLEHUB_CACHE"] = "./temp_cache"
    
    # Le dataset officiel attendu par ton projet
    dataset_name = "paultimothymooney/breast-histopathology-images" 
    downloaded_path = kagglehub.dataset_download(dataset_name)
    
    # On déplace les dossiers patients vers notre dossier final
    os.makedirs(dossier_final, exist_ok=True)
    
    # Le dataset contient un dossier 'IDC_regular_ps50_idx5' qui contient lui-même les dossiers patients
    source_dir = os.path.join(downloaded_path, "IDC_regular_ps50_idx5")
    if not os.path.exists(source_dir): # Parfois kagglehub extrait directement à la racine
        source_dir = downloaded_path
        
    for item in os.listdir(source_dir):
        shutil.move(os.path.join(source_dir, item), dossier_final)
        
    shutil.rmtree("./temp_cache", ignore_errors=True)
    print(f"🎉 Terminé ! Tes données sont prêtes dans : {dossier_final}")
else:
    print(f"✅ Le dataset est déjà présent dans : {dossier_final}")
```

```python
import os
import shutil
import kagglehub

# Définition des chemins
base_dir = "../data" # Mets "../data" si ton notebook est dans un sous-dossier
target_dir = os.path.join(base_dir, "ambarish")
temp_cache = "./temp_cache_kaggle"

# On vérifie si les données sont déjà prêtes
if os.path.exists(target_dir) and "benign" in os.listdir(target_dir):
    print(f"✅ Le dataset BreaKHis est déjà prêt et propre dans : {target_dir}")
else:
    print("📥 Téléchargement de BreaKHis en cours (environ 4 Go)...")
    
    # Force kagglehub à utiliser notre dossier temporaire
    os.environ["KAGGLEHUB_CACHE"] = temp_cache
    dataset_name = "ambarish/breakhis"
    
    # Téléchargement
    downloaded_path = kagglehub.dataset_download(dataset_name)
    
    # Création du dossier cible final
    os.makedirs(target_dir, exist_ok=True)
    
    print("🧹 Restructuration de l'arborescence...")
    
    # Parcours du dossier temporaire pour récupérer ce qui nous intéresse
    for root, dirs, files in os.walk(downloaded_path):
        
        # Déplacement des dossiers benign et malignant
        for d in dirs:
            if d in ["benign", "malignant"]:
                src = os.path.join(root, d)
                dest = os.path.join(target_dir, d)
                if not os.path.exists(dest):
                    shutil.move(src, dest)
                    print(f"  -> Dossier déplacé : {d}")
                    
        # Déplacement de Folds.csv
        for f in files:
            if f == "Folds.csv":
                src = os.path.join(root, f)
                dest = os.path.join(target_dir, f)
                if not os.path.exists(dest):
                    shutil.move(src, dest)
                    print(f"  -> Fichier déplacé : {f}")
    
    # Nettoyage du dossier temporaire
    shutil.rmtree(temp_cache, ignore_errors=True)
    
    print(f"🎉 Terminé ! Tout est rangé proprement dans : {target_dir}")
```
